Passos 1-6 - Ler o CSV e criar uma cópia de trabalho

```python
import pandas as pd
from IPython.display import display, Markdown
import numpy as np

# 1. Ler o conteúdo do CSV fornecido
# Ajuste o caminho do arquivo conforme necessário
# Como o separador é ponto e vírgula, usamos sep=";"
df = pd.read_csv("pico_web.csv", sep=";", engine="python", encoding="utf-8")

# 2. Atribuir os dados lidos a uma variável
# Já atribuímos à variável df

# 3. Verificar se os dados foram importados adequadamente

display(Markdown("**Informações gerais sobre o conjunto de dados:**"))
display(df.info())  # mostra tipos de dados, quantidade de entradas, etc.

# 4. Imprimir as primeiras N linhas
N = 5
display(Markdown(f"\nPrimeiras {N} linhas do arquivo:"))
display(df.head(N))

# 5. Imprimir as últimas N linhas
display(Markdown(f"\nÚltimas {N} linhas do arquivo:"))
display(df.tail(N))

# 6. Criar uma nova variável com uma cópia do conjunto de dados original
df_copia = df.copy()

# Verificação: mostrar que a cópia existe
display(Markdown("\nCópia do conjunto de dados criada com sucesso!"))
display(Markdown("\nExibindo a cópia do conjunto de dados:"))
display(df_copia)  
```

Passo 7 - Substituir NaN em Calories por 0 e verificar

```python
# 7. Substituir valores nulos da coluna 'Calories' por 0
df_copia['Calories'] = df_copia['Calories'].fillna(0)

# Imprimir o conjunto de dados para verificar a mudança
display(Markdown("\nConjunto de dados após substituir valores nulos em 'Calories' por 0:"))
display(df_copia)
display(df_copia.info())  # Verificar se os nulos foram substituídos
```

Passo 8 - Substituir nulos por "1900/01/01"

```python
# 8.Substitui valores nulos por '1900/01/01'
df_copia['Date'] = df_copia['Date'].fillna('1900/01/01')
display(Markdown("\nConjunto de dados após substituir valores nulos em 'Date' por '1900/01/01':"))
display(df_copia)
# Tentativa de conversão (gera erro)
df_copia['Date'] = pd.to_datetime(df_copia['Date'], format='%Y/%m/%d')
```

Passo 9 - Tentativa de conversão direta e substituição de '1900/01/01' por NaN

```python
# 1. Substituir '1900/01/01' por NaN
df_copia['Date'] = df_copia['Date'].replace('1900/01/01', np.nan)

# 2. Converter novamente para datetime
df_copia['Date'] = pd.to_datetime(df_copia['Date'])

# 3. Imprimir para verificar se a mudança foi aplicada
display(Markdown("### Conjunto de dados após substituir '1900/01/01' por NaN e converter para datetime:"))
display(df_copia.info())
display(df_copia)
```

Passo 10-11 - Tratar '20201226' e converter Date para datetime

```python
# 2. Corrigir especificamente o valor '20201226' transformando em datetime
df_copia['Date'] = df_copia['Date'].replace(
    '20201226', pd.to_datetime('20201226', format='%Y%m%d')
)

# 3. Converter novamente toda a coluna para datetime
df_copia['Date'] = pd.to_datetime(df_copia['Date'], errors='coerce')

# 4. Imprimir para verificar se a mudança foi aplicada
display(Markdown("### Conjunto de dados após corrigir '20201226' e converter toda a coluna:"))
display(df_copia)
display(df_copia.info())
```

Passo 12 - Remover os registros contento valores nulos

```python
df_copia = df_copia.dropna()
display(df_copia.info())
```

Passo 13 - DataFrame apos todas as Transformacões

```python
display(Markdown("### Conjunto de dados após remover registros com valores nulos:"))
display(df_copia)
```
