Microatividade 1: Descrever como ler um arquivo CSV usando a biblioteca Pandas (Python)

```python
import pandas as pd
from IPython.display import display

csv_path = "pico_web.csv"
df =pd.read_csv(csv_path, sep=';', engine='python', encoding='utf-8')

# Exibição formatada
display(df)
```

Microatividade 2: Descrever como criar um subconjunto de dados a partir de um conjunto existente usando a biblioteca Pandas (Python)

```python
import pandas as pd
from IPython.display import display

cols = ['Duration', 'Pulse', 'Calories']

missing = [c for c in cols if c not in df.columns]
if missing:
    raise KeyError(f"Colunas não encontradas no DataFrame: {missing}")

subset = df[cols].copy()
display(subset.head(10))
```

Microatividade 3: Descrever como configurar o número máximo de linhas a serem exibidas na visualização de um conjunto de dados usando a biblioteca Pandas (Python)

```python
import pandas as pd
from IPython.display import display

pd.options.display.max_rows = 9999

display(df)
```

Microatividade 4: Descrever como exibir as primeiras e últimas “N” linhas de um conjunto de dados usando a biblioteca Pandas (Python)

```python
# Em notebooks, display é mais legível; print funciona também
from IPython.display import display

print("Primeiras 10 linhas:")
display(df.head(10))

print("\nÚltimas 10 linhas:")
display(df.tail(10))

# Verificação rápida: número total de linhas
print("\nTotal de linhas no DataFrame:", df.shape[0])
```

Microatividade 5: Descrever como exibir informações gerais sobre as colunas, linhas e dados de um conjunto de dados usando a biblioteca Pandas (Python)

```python
import pandas as pd
from IPython.display import display

print("=== info() ===")
df.info()

print("\n=== shape ===")
print(df.shape)

print("\n=== Contagem de valores nulos por coluna ===")
display(df.isnull().sum())

print("\n=== Tipos de dados por coluna ===")
display(df.dtypes)

print("\n=== Uso de memória por coluna (bytes) ===")
display(df.memory_usage(deep=True))
```
